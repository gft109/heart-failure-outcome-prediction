### 2.3 Valores Faltantes

Nem todo `NA` é um dado perdido aleatoriamente (Missing At Random). Exames como gasometria e perfil lipídico costumam faltar em blocos, indicando que **o painel não foi pedido**. Como essa ausência carrega informação clínica (o médico julgou desnecessário), a etapa 2 (`02_selecao_atributos.ipynb`) cria indicadores de exame não pedido em vez de apenas imputar os valores.

A função a seguir, do modulo `eda.py`, calcula o percentual exato de valores nulos (ausentes) para cada uma das 166 colunas da base de pacientes e gera um gráfico de barras com a porcentagem de faltantes.

> Filtragem: Para não poluir o gráfico, a função exclui as variáveis que estão 100% preenchidas e ordena as restantes da menor para a maior taxa de ausência.

```python
eda.plot_missing_rate_by_variable(patients)
```

**Quanto falta em cada variável**: 
- Cada barra é uma variável com algum faltante. As linhas tracejadas marcam 20% e 50%, limiares usuais na literatura para separar faltantes esporádicos de exames que não foram pedidos. Há um grupo grande acima de 50% (gasometria, lipídios, ecocardiograma) e quase todo o resto abaixo de 20%.

**Interpretação para a modelagem**:
- Abaixo de 20%: Valores ausentes aqui (como em alguns exames básicos) costumam ser aleatórios e seguros para imputação (substituir o nulo pela mediana ou média da população).
- Acima de 50%: Barras longas que ultrapassam a segunda linha (como gasometria, lipoproteínas, ou respiratory.support. com 97,9% de ausência) indicam dados que não faltam ao acaso (Not Missing At Random). Imputar esses valores sem mais nada trataria "exame não pedido" como se fosse um resultado típico. Por isso, na etapa 2, toda variável com mais de 20% de faltantes ganha um indicador binário de "exame não pedido" (um por painel), pois a ausência carrega a informação clínica de que o médico não suspeitou daquela complicação. O valor medido é mantido ao lado do indicador e a seleção de atributos decide se ele ajuda.

### 2.3.1 Matriz de Valores Faltantes

Esta visualização comprova o padrão NMAR (Not Missing At Random). As faixas que atravessam o painel confirmam que não faltam pontos aleatórios de dados, mas sim painéis inteiros que não foram solicitados pelo médico.\
A função `plot_missing_matrix` gera um "mapa de calor" booleano que permite visualizar a estrutura e o comportamento dos dados ausentes em toda a base de pacientes.
O eixo horizontal representa as variáveis agrupadas, e o eixo vertical representa os 2.008 pacientes. Onde o dado está preenchido, a imagem fica na cor clara (fundo neutro); onde o dado é nulo/ausente, a imagem é preenchida por blocos azuis (cor da série).
- **Impacto na Modelagem**: Esse comportamento em bloco demonstra que a ausência ocorre porque o painel laboratorial inteiro não foi solicitado pelo médico. Como essa ausência indica uma decisão médica presuntiva (o médico julgou o paciente estável o suficiente para não precisar da gasometria, por exemplo), a imagem suporta metodologicamente a decisão de criar um indicador por painel ("exame não pedido") na etapa 2, em vez de apenas imputar os valores.

```python
eda.plot_missing_matrix(patients, variable_groups)
```

### 2.4 Variável Alvo: Óbito ou Readmissão em 6 meses

**Motivação**: um teste preliminar comparou vários desfechos candidatos (readmissão em 28 dias/3/6 meses, óbito, alta a pedido, tempo de internação, classe NYHA). O composto *óbito ou readmissão em 6 meses* é o endpoint mais usado em estudos de prognóstico de IC, tem as classes mais equilibradas e evita uma distorção da readmissão pura: quem morre após a alta não pode ser readmitido e seria contado como "sem evento".

**Definição**: `target = 1` se o paciente foi readmitido em qualquer janela registrada (28 dias, 3 ou 6 meses) **ou** morreu em até 6 meses; senão `target = 0`. Usar qualquer janela de readmissão (e não só a de 6 meses) cobre 3 registros inconsistentes em que há readmissão em 3 meses mas não em 6 (ver checagem abaixo).

O gráfico a seguir mostra a prevalência de cada desfecho disponível na base, para situar o alvo escolhido em relação às alternativas: desfechos de óbito são raros demais (< 3%) para modelar com segurança, enquanto readmissão e o composto têm classes suficientemente equilibradas.

A função `create_target()` do módulo `preprocessing.py` realiza o agrupamento destas variáveis.

**Composição do alvo e checagens de consistência**: Os desfechos são cumulativos no tempo (quem teve o evento em 28 dias deve tê-lo também em 3 e 6 meses); óbito e readmissão em 6 meses não se sobrepõem nesta base. Os óbitos durante a própria internação também entram no alvo, pois estão contidos em `death.within.6.months`.

```python
target = preprocessing.create_target(patients)
print(f"Total no alvo: {target.sum()} de {len(target)} ({target.mean():.1%})")

# Gera a tabela de resumo, o gráfico de barras (prevalência) e o detalhamento
eda.plot_outcome_prevalence(patients, target)
eda.print_outcome_details(patients, target)
```
